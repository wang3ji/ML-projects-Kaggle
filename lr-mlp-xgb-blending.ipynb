```python
test_df = pd.read_csv("/kaggle/input/competitions/playground-series-s6e3/test.csv")
test_ids = test_df["id"]  
```

```python
lr_prob = np.load("/kaggle/input/notebooks/jwang2025learning/lr-tuning/lr_best_predict.npy")
xgb_prob = np.load("/kaggle/input/notebooks/jwang2025learning/xgboost-tuning/xgb_best_predict.npy")
mlp_prob = np.load("/kaggle/input/notebooks/jwang2025learning/mlp-tuning/MLP_best_predict.npy")
```

```python
w_lr = 0.2
w_mlp = 0.5
w_xgb = 0.3

probs_blend = w_lr * lr_prob + w_mlp * mlp_prob + w_xgb * xgb_prob
```

```python
submission = pd.DataFrame({
    "id":    test_ids,
    "Churn": probs_blend
})

submission.to_csv("submission.csv", index=False)
print(f"\nSubmission saved — Churn distribution:\n{submission['Churn'].value_counts()}")
print(submission.head())
```
